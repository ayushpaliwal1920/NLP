```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm') # dict import
```

```python
doc = nlp("A Point of Sale (POS) system is the technology, combining software and hardware like registers, scanners, and card terminals")
```

```python
doc.text
```

```python
doc[-1] # indexing
```

```python
# pos_ : uses for 'course grianed' parts of speech

doc[0].pos_ # pos
```

```python
doc[1].pos_ # pos
```

```python
doc[2].pos_ # pos
```

```python
# tag_ : uses for 'fine grained 'parts of speech : detailed

doc[0].tag_
```

```python
doc[2].tag_
```

```python
doc[-1].tag_
```

```python
# for explaining ;

spacy.explain('NNP')
```

```python
spacy.explain('VB')
```

```python
for word in doc:
  print(word.text,"----->",word.pos_,word.tag_,spacy.explain(word.tag_))
```

It can compute detailed/minute differences

```python
doc2 = nlp(u"I read bookd on history")
for word in doc2:
  print(word.text, "------->" , word.pos_ , word.tag_ , spacy.explain(word.tag_))
```

```python
doc3 = nlp(u"I have read a book on history ")
for word in doc2:
  print(word.text, "------->" , word.pos_ , word.tag_ , spacy.explain(word.tag_))
```

```python
# Visualization :

from spacy import displacy

doc6 = nlp(u"The quick brown fox jumped over the lazy dog")

options = {
    'distance' : 80,
    'compact'  : True,
    'color' : '#fff',
    'bg' : '#000'
}
```

```python
displacy.render(doc6,style = 'dep' , jupyter= True)
```

```python
displacy.render(doc6,style = 'dep' , jupyter= True,options=options)
```
